```python
#Write a program to smooth the input image using:
#1. Averaging filter (un-weighted)
import cv2
import matplotlib.pyplot as plt
import os

image_path = "image4.png"
if not os.path.exists(image_path):
    # Defaulting to the available image if image4.png is not found
    image_path = "image.jpg"

img = cv2.imread(image_path)

if img is None:
    raise FileNotFoundError(f"Could not load image: {image_path}. Please check the file path.")

# Convert BGR to RGB for correct displaying in matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

size = int(input("Enter filter size (odd number): "))

print("1. Zero Padding")
print("2. Replication Padding")
print("3. Reflect Padding")

p = int(input("Enter padding method: "))

if p == 1:
    border = cv2.BORDER_CONSTANT
elif p == 2:
    border = cv2.BORDER_REPLICATE
else:
    border = cv2.BORDER_REFLECT

output = cv2.blur(img_rgb, (size, size), borderType=border)

plt.imshow(output)
plt.axis("off")
plt.show()
```

```python
#2. Weighted filter given by h(x,y)= max(|x|,|y|)
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = "image4.png"
if not os.path.exists(image_path):
    # Defaulting to the available image if image4.png is not found
    image_path = "image.jpg"

img = cv2.imread(image_path)

if img is None:
    raise FileNotFoundError(f"Could not load image: {image_path}. Please check the file path.")

# Convert BGR to RGB for correct displaying in matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

size = int(input("Enter filter size (odd number): "))

print("1. Zero Padding")
print("2. Replication Padding")
print("3. Reflect Padding")

p = int(input("Enter padding method: "))

if p == 1:
    border = cv2.BORDER_CONSTANT
elif p == 2:
    border = cv2.BORDER_REPLICATE
else:
    border = cv2.BORDER_REFLECT

n = size // 2

kernel = np.zeros((size, size), dtype=float)

for x in range(-n, n + 1):
    for y in range(-n, n + 1):
        kernel[x+n, y+n] = max(abs(x), abs(y))

kernel = kernel / np.sum(kernel)

output = cv2.filter2D(img_rgb, -1, kernel, borderType=border)

plt.imshow(output)
plt.axis("off")
plt.show()
```

```python
#3. Gaussian filter
import cv2
import matplotlib.pyplot as plt
import os

image_path = "image4.png"
if not os.path.exists(image_path):
    # Defaulting to the available image if image4.png is not found
    image_path = "image.jpg"

img = cv2.imread(image_path)

if img is None:
    raise FileNotFoundError(f"Could not load image: {image_path}. Please check the file path.")

# Convert BGR to RGB for correct displaying in matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

size = int(input("Enter filter size (odd number): "))
sigma = float(input("Enter sigma value: "))

print("1. Zero Padding")
print("2. Replication Padding")
print("3. Reflect Padding")

p = int(input("Enter padding method: "))

if p == 1:
    border = cv2.BORDER_CONSTANT
elif p == 2:
    border = cv2.BORDER_REPLICATE
else:
    border = cv2.BORDER_REFLECT

output = cv2.GaussianBlur(img_rgb, (size, size), sigma,
                          borderType=border)

plt.imshow(output)
plt.axis("off")
plt.show()
```
