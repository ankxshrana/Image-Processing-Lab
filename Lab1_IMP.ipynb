```python
import cv2
import urllib.request
import os
from google.colab.patches import cv2_imshow
from IPython.display import clear_output
import time
```

**STEP 0: Download Sample Data from Online Libraries**

```python
print("Downloading sample image and video...")

IMAGE_URL = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
VIDEO_URL = "https://raw.githubusercontent.com/intel-iot-devkit/sample-videos/master/car-detection.mp4"

urllib.request.urlretrieve(IMAGE_URL, "sample_image.jpg")
urllib.request.urlretrieve(VIDEO_URL, "sample_video.mp4")

print("Download complete!\n")
```

**PART 1: IMAGE OPERATIONS**

```python
# a. Read, display, and write an image
img = cv2.imread("sample_image.jpg")  # Read
print("Original Image:")
cv2_imshow(img)                       # Display in Colab
cv2.imwrite("saved_copy.jpg", img)    # Write / Save
```

```python
# b. Convert the image into different formats (PNG, TIFF)
cv2.imwrite("output_image.png", img)
cv2.imwrite("output_image.tiff", img)
```

```python
# c. Perform image resizing (200x200 pixels)
resized_img = cv2.resize(img, (200, 200))
print("\nResized Image (200x200):")
cv2_imshow(resized_img)
cv2.imwrite("resized_image.jpg", resized_img)
```

```python
# d. Convert a colored image into a grayscale image
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("\nGrayscale Image:")
cv2_imshow(gray_img)
cv2.imwrite("grayscale_image.jpg", gray_img)
```

```python
print("\n--- Image operations completed! ---\n")
```

**PART 2: VIDEO OPERATIONS**

```python
# e. Read and display a video frame-by-frame in Colab
cap = cv2.VideoCapture("sample_video.mp4")

frame_count = 0
print("Playing video (showing first 30 frames as a demo)...")

while cap.isOpened() and frame_count < 30:
    ret, frame = cap.read()
    if not ret:
        break

    # Display frame in notebook
    clear_output(wait=True)
    cv2_imshow(frame)
    time.sleep(0.03)  # Small delay to simulate playback speed

    frame_count += 1

cap.release()
```

```python
# f. Extract frames from a video
os.makedirs("extracted_frames", exist_ok=True)
cap = cv2.VideoCapture("sample_video.mp4")

saved_frames = 0
total_frames = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Save every 15th frame
    if total_frames % 15 == 0:
        cv2.imwrite(f"extracted_frames/frame_{total_frames}.jpg", frame)
        saved_frames += 1

    total_frames += 1

cap.release()

print(f"\n--- Video operations complete! Extracted {saved_frames} frames to 'extracted_frames/' folder. ---")
```
